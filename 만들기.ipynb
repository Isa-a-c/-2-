직사각형·원형 막에는 **2차원 애니메이션**을 제공합니다. 시간 전체에 고정된 색상 범위를 사용하므로 진폭 변화를 비교할 수 있습니다. 초기 함수와 경계조건의 불일치, 성긴 시간 표본은 안내합니다.

### 입력과 가정

- 무한 현·막대: `x`를 사용하며 화면 밖에서도 정의된 함수를 입력하십시오. 화면 끝에서 반사되거나 온도가 고정되는 모델이 아닙니다.
- 직사각형 막: `x,y,ell,height`를 사용합니다. `ell,height`는 가로·세로 길이입니다.
- 원형 막: `r,theta,ell`을 사용합니다. `ell`은 반지름이며 중심 r=0의 값은 각도에 무관해야 합니다.
- 원판 퍼텐셜: 경계 함수 `h(theta)`를 입력합니다. 구 퍼텐셜은 `h(mu)`, `mu=cos(φ)`를 입력합니다.
- 반무한 현: 경계 함수 `h(t)`를 입력합니다. 초기 변위·속도는 0이며 반무한 열전도는 상수 계단 경계만 지원합니다.

막은 일정 속력·고정 가장자리·무감쇠·무외력 모델입니다. 퍼텐셜은 중심에서 유한한 내부 Dirichlet 문제이며 일반 3차원·외부 영역·Neumann·Robin 조건까지 지원하지는 않습니다. Laplace PDE 화면도 임의 PDE를 자동으로 푸는 범용 기호 해석기는 아닙니다.

### 코드와 검증

`core/chapter12.py`는 초기값 공식·열핵 구적·막의 모드 계산·퍼텐셜을 담당합니다. `topics/chapter12.py`는 입력 검증·표·그래프를, `topics/chapter12_lessons.py`는 개념과 코드 해설을 담당합니다. `ui/animation.py`는 기존 1차원 재생과 새 2차원 재생을 함께 지원합니다.

`tests/test_chapter12.py`는 진행파·Gaussian 확산·단일 막 모드·에너지 보존·퍼텐셜·인과성·잘못된 입력을 검증합니다. 사다리꼴 적분의 유한 격자 오차는 해석해 비교 허용오차에 반영하고 구의 퍼텐셜에서 격자 세분화 수렴을 확인합니다. `만들기.ipynb`에도 12장 순서의 존댓말 설명·실행 예제·검산을 추가하였습니다.

연속 Fourier 변환은 교재 56쪽의 **F(ω)=1/√(2π) ∫f(x)e^(−iωx)dx**를 사용합니다. 역변환도 1/√(2π)를 사용하며 ω는 각주파수입니다. 기존 DFT의 Hz 표기와 구분하십시오.

11.5는 교재 부호 `(p y′)′+(q+λr)y=0`를 사용합니다. 상수 p,r>0에서 λ=(pκ²−q)/r입니다. 가변계수·임의 Robin 조건은 아직 지원하지 않습니다. 11.6의 N은 기저 개수이므로 Legendre에서는 0~N−1차를 포함하며, Legendre 구간은 L과 관계없이 [-1,1]입니다.

11.7~11.10은 공간·주파수 적분을 유한 범위로 절단합니다. 수치 결과는 무한구간 변환의 존재를 판정하지 않습니다. 표본 수와 절단 범위를 따로 변경하십시오. 11.10은 표의 대표 공식 11개를 지원하며 특이·조건부 수렴 적분을 포함한 표 전체의 자동 계산기는 아닙니다.

## 설치 및 실행

Python 3.11 이상을 권장합니다. 프로젝트 루트에서 다음 명령을 실행하십시오.

```powershell
python -m venv .venv
.venv\Scripts\python.exe -m pip install -r requirements.txt

```

환경이 준비되어 있다면 `실행.bat`을 더블클릭하십시오. Qt의 공식 Python 바인딩인 PySide6 창이 열리며 주력 실행 경로는 `main.py`입니다. 테스트와 노트북 셀 검사는 `pip install -r requirements-dev.txt`로 개발 의존성을 설치하십시오.

배포판은 `dist/EngineeringMathStudio/EngineeringMathStudio.exe`를 실행하십시오. `_internal`을 포함한 전체 폴더가 필요합니다. Python 설치는 필요하지 않습니다.

개발 환경에서 배포판을 다시 만들려면 `requirements-build.txt`를 설치한 뒤 `build_windows.ps1`, `python scripts/check_portable.py`, `python scripts/package_release.py` 순서로 실행하십시오. 검사 범위는 `VALIDATION.md`에 기록되어 있습니다.

## 사용 방법

1. 왼쪽 위에서 학습 주제를 선택하십시오.
2. 예제를 선택하거나 수식과 매개변수를 입력하십시오. 정수 입력은 슬라이더와 숫자 입력이 동기화됩니다.
3. **계산하기**를 누르십시오. 계산 중에도 창을 조작하거나 계산을 취소할 수 있습니다.
4. 그래프 탭, 수치 표, 오차 및 안내문을 확인하십시오. 그래프 도구 모음으로 확대하거나 이미지를 저장할 수 있습니다.
5. **변화 과정 재생** 버튼이 활성화된 주제에서는 차수 또는 시간에 따른 변화를 관찰하십시오. 12장의 막은 2차원으로 재생됩니다. 새 11장 모듈과 PDE 기호 검산·정적 퍼텐셜·ODE·라플라스 변환은 정적 그래프를 제공합니다.
6. 파일 메뉴에서 입력 설정(JSON)과 계산 결과(CSV ZIP)를 저장하십시오. 결과 저장은 마지막 성공 계산 당시의 입력을 사용합니다.

설정을 바꾸어도 기존 결과는 자동으로 바뀌지 않습니다. 변경 안내가 나타나면 다시 계산하십시오. **계산 설정** 탭에는 결과를 만든 당시의 입력이 표시됩니다. 계산 중 입력을 수정한 경우에도 완료 결과와 현재 입력의 차이를 표시합니다. **학습 설명** 탭은 계산 전에도 읽을 수 있으며 정의·가정·계산 과정·결과 해석·관찰 과제를 제공합니다.

계산은 별도 프로세스에서 수행되며 취소 시 해당 프로세스를 종료합니다. 마지막으로 성공한 푸리에 계수의 캐시를 다음 프로세스에 전달합니다. 함수·L·적분 표본 수가 같으면 계수를 재사용하고, 더 큰 N에 필요한 차수만 추가 적분합니다. N을 줄여도 캐시에 저장된 높은 차수 계수는 유지합니다. 함수·L·표본 수를 바꾸면 새로 적분하며 실패·취소된 작업은 성공 캐시를 교체하지 않습니다. 안내문에 재사용·추가 적분한 차수를 표시합니다.

## 수식·매개변수 입력

수식 함수의 조합을 직접 입력할 수 있습니다. 곱셈은 `2*x`처럼 명시하고 거듭제곱은 `x**2` 또는 `x^2`로 입력하십시오. 푸리에에서는 x, PDE 초기조건에서는 x·ell, ODE 외력·라플라스 정변환에서는 t, 역변환에서는 s를 사용합니다.

수식을 직접 입력하는 주제에는 매개변수 입력을 제공합니다. 11.10에서는 선택한 공식의 a·b를 숫자로 지정합니다. `A=2; w=3*pi`처럼 세미콜론으로 구분한 독립적인 유한 실수 상수를 선언하고, 수식에서 `A*sin(w*x)`처럼 사용하십시오. 매개변수는 최대 16개이며 다른 매개변수를 참조할 수 없습니다. x·t·s·ell·pi·E·I와 함수 이름은 재정의할 수 없습니다. 수식 외의 길이·질량 등 숫자 위젯에는 숫자를 직접 입력하십시오. 설정 JSON과 결과 ZIP에는 선언한 매개변수도 저장됩니다.

| 범주 | 예시 |
|---|---|
| 삼각·역삼각 | sin, cos, tan, asin, acos, atan, atan2, arcsin, arccos, arctan |
| 쌍곡·역쌍곡 | sinh, cosh, tanh, asinh, acosh, atanh |

`sinc(x)`는 sin(x)/x이고 x=0에서 1입니다. NumPy의 정규화된 sinc와 구별하십시오. 구간별 예제는 `Piecewise((1, And(t>=1, t<2)), (0, True))`입니다. 전체 지원 이름은 앱의 학습 설명 탭에 표시됩니다.

## 푸리에 급수의 정의

함수는 구간 $[-L,L]$에서 주어지고 주기는 $2L$로 가정합니다.

$$f(x)\sim a_0+\sum_{n=1}^{\infty}\left[a_n\cos\frac{n\pi x}{L}+b_n\sin\frac{n\pi x}{L}\right]$$

$$a_0=\frac{1}{2L}\int_{-L}^{L}f(x)\,dx$$
$$a_n=\frac{1}{L}\int_{-L}^{L}f(x)\cos\frac{n\pi x}{L}\,dx,\qquad b_n=\frac{1}{L}\int_{-L}^{L}f(x)\sin\frac{n\pi x}{L}\,dx$$

상수항은 $a_0$이며 $a_0/2$가 아닙니다. `numpy.trapezoid()`로 계수를 적분합니다. `n`은 개별 조화파의 번호, `N`은 부분합의 마지막 차수입니다. `num_points`는 적분 표본 수이며 `M`은 별도의 DFT 표본 수입니다.

현재 부분합, 여러 N의 비교, MSE·RMSE·최대 절대 오차, 계수 표와 스펙트럼을 제공합니다. 추가 분석에는 Gibbs 확대, 진폭·위상, DFT·FFT·급수 비교, Aliasing 실험, 적분 안정성, N별 오차 곡선이 있습니다. 비교 N을 비워도 계산할 수 있습니다. 애니메이션은 0차부터 계산한 현재 N까지 재생합니다.

### 급수, 연속 변환, DFT 및 FFT의 차이

푸리에 급수는 주기 함수의 이산적인 조화파 계수를 구합니다. 연속 푸리에 변환은 비주기 함수를 연속 주파수의 관점에서 다룹니다. 예를 들어 $F(\omega)=\int_{-\infty}^{\infty}f(x)e^{-i\omega x}dx$라는 정의를 사용할 수 있습니다. 이 앱은 **연속 푸리에 변환의 무한 구간 적분을 구현하지 않습니다**.

DFT는 유한 표본 $y_j$에 대해 $X_k=\sum_{j=0}^{M-1}y_j e^{-2\pi i jk/M}$를 계산합니다. FFT는 같은 DFT를 효율적으로 계산하는 알고리즘입니다. 앱에서는 직접 DFT와 `numpy.fft.fft`를 비교합니다. 구간 끝점은 중복하지 않으며, 시작점 $-L$의 위상을 보정한 계수는 $c_k=X_k e^{i\pi k}/M$입니다. 주파수 단위는 cycles/x입니다. DFT CSV에는 원래 복소 변환값과 보정 계수, 양측 크기 및 위상이 함께 저장됩니다. 거의 0인 크기의 위상은 공란입니다.

## 열방정식과 푸리에 급수의 연결

길이 $\ell$인 막대에서 $u_t=\alpha u_{xx}$를 계산합니다. 경계조건은 $u(0,t)=T_0$, $u(\ell,t)=T_1$입니다. 정상 상태 $g(x)=T_0+(T_1-T_0)x/\ell$를 먼저 분리합니다.

$$u(x,t)\approx g(x)+\sum_{n=1}^{N}b_n\sin\frac{n\pi x}{\ell}\exp\left[-\alpha\left(\frac{n\pi}{\ell}\right)^2t\right]$$
$$b_n=\frac{2}{\ell}\int_0^{\ell}[u(x,0)-g(x)]\sin\frac{n\pi x}{\ell}\,dx$$

잔차 함수를 기확장하여 기존 푸리에 계수 함수를 재사용합니다. 입력에는 `x`와 `ell`을 사용할 수 있습니다. 시간별 온도 곡선, 시공간 히트맵, 사인 계수, 잔차 에너지, 시간 애니메이션을 제공합니다. 초기 함수와 경계 온도가 맞지 않으면 안내문을 표시하며, 해의 끝점은 경계조건으로 고정합니다. 열확산계수의 차원은 길이²/시간입니다.

이 방법은 일정한 열확산계수와 시간에 무관한 Dirichlet 경계조건을 가정합니다. 일반적인 PDE 수치 해석기나 유한차분법은 아닙니다.

### 열방정식 정확도 비교

**초기 재구성** 탭은 t=0에서 초기 함수와 유한 사인 급수의 차이를 표시합니다. 시간에 따른 온도 감소와 재구성 오차를 구분하십시오. 수치 표의 **정확도 비교**에는 기준 결과, 적분 표본 수를 `2*num_points-1`로 바꾼 결과, 차수를 `min(2*N,200,num_points-2)`로 바꾼 결과가 있습니다. 비교는 공통 2049개 평가점에서 수행합니다. 표본 수를 바꿀 때는 N을, N을 바꿀 때는 적분 표본 수를 유지합니다. 초기 재구성 RMSE, 기준과의 초기·최종 시각 최대 차이를 함께 제공합니다. 두 근사 해의 차이가 작아도 참해에 대한 오차가 작다고 보장하지는 않습니다.

## 파동방정식과 열방정식 비교

일정한 속력 c의 고정단 현에서 $u_{tt}=c^2u_{xx}$, $u(0,t)=u(\ell,t)=0$을 계산합니다. 초기 변위 $f(x)$와 초기 속도 $v(x)$의 사인 계수를 각각 $A_n,V_n$이라 하면 다음과 같습니다.

$$\omega_n=cn\pi/\ell,\qquad u(x,t)\approx\sum_{n=1}^{N}\left[A_n\cos(\omega_nt)+\frac{V_n}{\omega_n}\sin(\omega_nt)\right]\sin\frac{n\pi x}{\ell}$$

초기 속도 계수는 시간 미분 조건을 맞추기 위해 $\omega_n$으로 나눕니다. 초기 변위·속도에서 x와 ell을 사용할 수 있습니다. 시간별 변위, 시공간 히트맵, 계수·에너지 표 및 애니메이션을 제공합니다. 감쇠·외력·시간에 따라 바뀌는 경계조건은 구현하지 않았습니다.

단위 선밀도의 에너지는 $E=\frac12\int_0^\ell(u_t^2+c^2u_x^2)dx$이며 모드의 직교성을 이용해 계산합니다. 무감쇠 모형에서 일정해야 하지만 에너지 보존만으로 급수 절단 오차를 판단할 수는 없습니다.

**열·파동 비교** 탭에서는 동일한 초기 모양과 0 경계조건의 열방정식을 함께 계산합니다. 비교 α는 열 결과에만 사용됩니다. 열은 감쇠하고 파동은 진동하는 차이를 공통 세로축 범위로 관찰하십시오. 두 방정식의 u는 서로 다른 물리량이므로 모양의 수학적 비교입니다. 파동에만 초기 속도가 적용됩니다.

시간 표본은 출력 간격이며 수치 시간 적분의 간격이 아닙니다. 최고 차수 모드의 한 주기당 출력이 10개 미만이면 안내합니다. 작은 고차 계수까지 포함한 보수적 안내이므로 필요하면 시간 범위를 줄이거나 표본 수를 늘리십시오.

## 상미분방정식: 질량–스프링–댐퍼

$$m y''+c y'+ky=F(t),\qquad y(0)=y_0,\quad y'(0)=v_0$$

질량 m>0, 강성 k>0, 감쇠 c≥0인 선형계입니다. 외력은 없음·상수·사인·지연 계단 또는 **직접 입력**을 선택할 수 있습니다. 직접 입력은 t만의 함수이며 y나 y′에 의존하는 비선형 ODE는 지원하지 않습니다. 외력 각주파수의 단위는 rad/s입니다.

고유 각주파수는 $\omega_n=\sqrt{k/m}$, 감쇠비는 $\zeta=c/(2\sqrt{mk})$입니다. 무감쇠·부족감쇠·임계감쇠·과감쇠 및 공진 예제를 제공합니다. 시간 응답·위상평면·에너지·외력 그래프에서 결과를 확인하십시오.

`solve_ivp`로 변위·속도를 적분합니다. 기본 외력 네 종류의 독립 기준 해는 외력을 포함한 증강 상태행렬의 지수로 계산합니다. 직접 입력은 DOP853과 세분화 RK45의 두 수치해를 비교하며 해석해 비교가 아닙니다. 출력 표본 수는 내부 적분 횟수와 다르며, 큰 계수 비율과 시간 범위로 작업 규모가 과도해지면 입력 조정을 요청합니다. 수치 적분 API의 의미는 [SciPy 공식 문서](https://docs.scipy.org/doc/scipy/reference/generated/scipy.integrate.solve_ivp.html)를 참고하십시오.

직접 입력 예시는 `A*exp(-t)*sin(w*t)`, 매개변수 `A=2; w=3`입니다. **최대 내부 간격**으로 시간 적분 간격의 상한을 설정하십시오. Heaviside·Piecewise·sign·abs에 포함된 선형 조건의 경계는 자동 분할합니다. **분할 시각**에 `1; 2` 또는 `tau; 2*tau`처럼 경계를 추가할 수 있습니다. 매우 좁은 펄스·빠른 진동·비선형 조건의 경계는 놓칠 수 있으며 두 수치해의 일치가 정확도 보장은 아닙니다. 출력 표본이 없는 짧은 분할 구간도 적분하여 다음 구간으로 상태를 전달합니다.

직접 입력의 기본 옵션은 **수치 계산만**입니다. **라플라스 변환도 계산**을 선택하면 기호 변환을 추가로 시도합니다. 변환 실패·미평가 결과가 있어도 수치 결과를 유지하고 **계산 가능 여부** 표에 상태를 기록합니다. 전달 가능한 닫힌 변환식을 얻으면 라플라스 연결 버튼이 활성화됩니다. 기호 계산이 오래 걸려 사용자가 작업을 취소하면 해당 작업의 새 결과는 제출되지 않습니다.

에너지는 $E=mv^2/2+ky^2/2$, 변화율은 $E'=Fv-cv^2$입니다. 감쇠가 있어도 외력이 에너지를 공급하면 에너지가 증가할 수 있습니다. 무감쇠 공진 예제는 초기조건 0에서 외력 각주파수와 고유 각주파수를 같게 설정합니다.

## 라플라스 변환과 ODE 연결

$$F(s)=\int_{0^-}^{\infty}e^{-st}f(t)dt$$

정변환 모드에서는 t의 함수, 역변환 모드에서는 s의 식을 입력하십시오. 기호 계산 탭에서 변환식을 선택·복사할 수 있고 CSV에도 저장됩니다. 정변환은 변환식·수렴 반평면·추가 조건을 함께 표시합니다. 역변환은 인과적 단측 해석을 사용하며 일반적인 양측 변환의 모든 수렴 영역을 열거하지 않습니다. [SymPy 공식 문서](https://docs.sympy.org/latest/modules/integrals/integrals.html)의 변환 함수와 미평가 결과 처리 방식을 따릅니다.

ODE에서 초기조건을 대입한 식은 다음과 같습니다.

$$Y(s)=\frac{F(s)+m(sy_0+v_0)+cy_0}{ms^2+cs+k}$$

ODE의 **이 ODE 결과를 라플라스 역변환으로 확인** 버튼을 누르면 마지막 성공 결과의 Y(s)가 라플라스 화면으로 전달됩니다. 현재 입력을 수정한 상태라면 ODE를 다시 계산한 후 연결하십시오.

시간 그래프는 t>0에서만 표시하여 Heaviside(0)의 관례와 초기값의 오른쪽 극한을 구별합니다. `1`의 역변환처럼 DiracDelta를 포함하는 결과는 일반 함수 그래프로 그리지 않습니다. 미평가 변환은 식 자체를 표시하며, 수치 배열로 평가할 수 없는 경우에도 기호 결과를 보존합니다. 기호 계산은 오래 걸릴 수 있으며 계산 취소 버튼으로 중단할 수 있습니다.

# 파동방정식: 모드별 진동

## 1. 정의
u_tt = c² u_xx는 현의 변위가 전파되는 과정을 나타냅니다. 초기 변위 f(x)와 초기 속도 v(x)가 모두 필요합니다.

## 2. 가정과 경계조건
일정한 양수 파동 속력 c, 고정단 u(0,t)=u(ell,t)=0, 외력과 감쇠가 없는 1차원 현을 가정합니다. c의 단위는 길이/시간입니다. 초기 변위와 초기 속도의 끝점도 0이면 경계조건과 양립합니다.

## 3. 계산 과정
Aₙ과 Vₙ은 각각 초기 변위와 초기 속도의 사인 계수입니다. ωₙ=cnπ/ell로 두면 다음과 같습니다.

qₙ(t)=Aₙ cos(ωₙt)+(Vₙ/ωₙ)sin(ωₙt)

u(x,t) ≈ Σ qₙ(t) sin(nπx/ell)

초기 속도 계수 Vₙ을 ωₙ으로 나누는 이유는 시간 미분값이 t=0에서 Vₙ이 되어야 하기 때문입니다. 시간은 수치 적분하지 않고 위 식에 직접 대입합니다.

## 4. 에너지와 출력 표본
단위 선밀도에서 E=(1/2)∫[u_t²+c²u_x²]dx이며, 직교성을 이용하면 E=ell/4 Σ[q′ₙ²+ωₙ²qₙ²]입니다. 이상적인 무감쇠 모형에서는 이 값이 일정합니다. 에너지 일정성만으로 공간 급수의 절단 오차가 작다고 판단할 수는 없습니다.

시간 표본이 부족하면 화면에서 진동을 놓칠 수 있습니다. 최고 차수 모드의 주기와 출력 간격을 비교한 안내를 확인하십시오. 높은 차수 계수가 거의 0이면 안내는 보수적일 수 있습니다.

## 5. 열·파동 비교
동일한 초기 모양과 0 경계조건으로 열방정식도 계산합니다. 열은 감쇠하며 파동은 진동합니다. 파동에만 초기 속도가 추가됩니다. 두 그림의 u는 서로 다른 물리량을 나타내므로 모양과 시간 변화의 수학적 비교로 해석하십시오. 비교 α는 열 결과에만 영향을 줍니다.

기본 예제에서 c=1, ell=1일 때 한 주기는 2입니다. 마지막 시간을 0.5로 바꾸어 초기 사인 변위가 0을 지나는지 확인하십시오. 초기 변위를 0, 초기 속도를 사인 함수로 바꾸어 위상 차이를 관찰하십시오.

1. `topics/wave.py`의 `inputs`는 초기 변위·속도·속력·길이·차수·표본을 선언합니다.

# 열방정식: 모드별 감쇠

## 1. 정의
u_t = α u_xx는 공간적으로 고르지 않은 온도가 확산되는 과정을 나타냅니다.

## 2. 가정과 경계조건
길이 ell, 일정한 양수 α, 고정된 양 끝 온도 T₀·T₁, 내부 열원이 없는 1차원 막대를 가정합니다. α의 단위는 길이²/시간입니다. 초기 함수에서 x와 ell을 사용할 수 있습니다.

## 3. 계산 과정
정상 상태 g(x)=T₀+(T₁−T₀)x/ell을 분리합니다. 초기 잔차 u(x,0)−g(x)를 사인 급수로 전개합니다.

bₙ = (2/ell) ∫[0,ell] [u(x,0)−g(x)] sin(nπx/ell) dx

u(x,t) ≈ g(x)+Σ bₙ sin(nπx/ell) exp[−α(nπ/ell)²t]

각 모드의 시간 계수는 직접 계산합니다. 시간 표본 수는 출력 간격을 정하며, 시간 적분기의 정확도를 정하는 값이 아닙니다.

## 4. 정확도와 시간 변화 구분
초기 재구성 그래프는 t=0에서 유한 급수가 입력을 얼마나 재현하는지 보여줍니다. 온도가 시간에 따라 줄어드는 것은 확산의 결과이며 초기 재구성 오차와 다릅니다.

수치 표의 정확도 비교에서는 같은 2049개 평가점에서 기준 결과, 적분 표본 수를 2배−1로 늘린 결과, 차수를 늘린 결과를 비교합니다. 차수 비교는 최대 200차이며 공간 표본 수보다 작게 제한합니다. N 변경 시 적분 표본 수를 유지합니다. 최종 시각의 차이는 두 근사 해끼리의 차이이며 참해와의 오차는 아닙니다.

초기 함수가 끝점 온도와 맞지 않으면 경계에서 초기 재구성 오차가 남습니다. 경계값 자체는 지정한 온도로 고정됩니다.

# 상미분방정식과 감쇠진동

## 1. 정의와 가정
m y″+c y′+k y=F(t)를 계산합니다. m>0, k>0, c≥0이며 질량·강성·감쇠는 일정합니다. 변위와 속도를 초기조건으로 지정합니다. 현재 외력은 없음·상수·사인·지연 계단 중에서 선택합니다. 임의의 비선형 ODE는 지원하지 않습니다.

## 2. 감쇠 분류
고유 각주파수는 ωₙ=√(k/m), 감쇠비는 ζ=c/(2√(mk))입니다. ζ=0은 무감쇠, 0<ζ<1은 부족감쇠, ζ=1은 임계감쇠, ζ>1은 과감쇠입니다. 이 분류는 동차계의 특성이며 외력을 주면 응답 형태가 달라집니다.

## 3. 계산 과정
y′=v, v′=(F−cv−ky)/m의 연립 1계 식을 solve_ivp로 계산합니다. 출력 표본과 내부 적분 간격은 서로 다릅니다. 계단 시작점에서 적분을 분할하여 외력의 변화를 놓치지 않도록 합니다. 독립 기준 해는 외력을 포함한 선형 상태행렬의 지수로 계산하며 수치해와 비교합니다.

## 4. 위상평면과 에너지
위상평면의 가로축은 변위, 세로축은 속도입니다. E=mv²/2+ky²/2이며 E′=Fv−cv²입니다. 외력 없는 무감쇠계에서는 보존되고, 감쇠계에서는 감소합니다. 외력이 있으면 에너지가 증가할 수도 있습니다.

## 5. 라플라스 연결
L[y′]=sY−y₀, L[y″]=s²Y−s y₀−v₀를 대입하면 다음을 얻습니다.

Y(s)=[F(s)+m(s y₀+v₀)+c y₀]/(m s²+c s+k)

기호 계산 탭에서 F(s), Y(s), 초기조건이 반영된 식을 확인하십시오. 연결 버튼은 마지막 성공 결과의 Y(s)를 역변환 화면으로 전달합니다. 입력을 수정했다면 먼저 다시 계산하십시오.

## 6. 관찰 과제
감쇠비를 바꾸어 시간 응답과 위상평면을 비교하십시오. 무감쇠·초기조건 0에서 사인 외력 각주파수를 ωₙ으로 맞추어 진폭 증가를 관찰하십시오. 양의 감쇠에서는 공진 증폭이 유한하며 변위 진폭 최대 주파수가 항상 ωₙ과 같지는 않습니다.


## ODE 코드별 역할

`inputs`는 질량·감쇠·강성·초기조건·외력을 선언합니다. `compute`의 `derivative`는 변위와 속도를 두 개의 1계 식으로 반환합니다. `solve_ivp`의 내부 적분과 출력 표본 수를 구분하십시오. 계단 외력은 시작점에서 구간을 나눕니다.

외력을 포함한 증강 상태행렬의 지수로 독립 기준 해를 계산합니다. 같은 적분 함수를 두 번 호출하는 검사가 아닙니다. `figures`는 결과 배열만 받아 시간 응답·위상평면·에너지를 그립니다. 감쇠계수를 0, 0.4, 4, 6으로 바꾸어 예상과 비교하십시오.


````python
from engineering_math.topics.ode import OdeTopic, FORCES
ode = OdeTopic()
ode_result = ode.compute({'damping': 0.4})
print(ode_result.data['regime'], ode_result.metrics)
for name, figure in ode.figures(ode_result).items():
    print(name)
    show(figure)
````

## 공진 해석해와 비교

m=1, c=0, k=4, 초기조건 0, 외력 sin(2t)이면 y(t)=sin(2t)/8−t cos(2t)/4입니다. 시간에 비례하는 진폭 항을 확인하십시오.


````python
resonance = ode.compute({'damping': 0., 'position': 0., 'mode': FORCES[2]})
ode_t = resonance.data['time']
exact_resonance = np.sin(2*ode_t)/8-ode_t*np.cos(2*ode_t)/4
print('공진 해석해 최대 차이:', np.max(np.abs(resonance.data['values'][0]-exact_resonance)))
np.testing.assert_allclose(resonance.data['values'][0], exact_resonance, atol=1e-8)
````

# 단측 라플라스 변환과 역변환

## 1. 정의
F(s)=∫[0⁻,∞] exp(−st)f(t)dt입니다. s는 복소수 변수이며 변환식뿐 아니라 수렴 영역도 중요합니다. 정변환에서는 f(t), 역변환에서는 F(s)를 입력하십시오.

## 2. 기호 계산과 조건
SymPy가 반환하는 정변환식, Re(s)>a 형태의 수렴 반평면 및 추가 조건을 함께 표시합니다. 닫힌 형태로 계산되지 않은 변환은 미평가 식으로 남겨 표시합니다. 계산이 오래 걸리면 취소할 수 있습니다.

## 3. 역변환의 해석
역변환은 인과적 단측 해석을 사용합니다. 같은 대수식이라도 다른 수렴 영역을 갖는 일반적인 양측 변환을 모두 열거하는 화면은 아닙니다. 입력에는 s 하나만 사용합니다.

## 4. 그래프와 분포
시간 그래프는 t>0에서 표시하여 Heaviside(0)의 관례와 초기값의 오른쪽 극한을 혼동하지 않도록 합니다. t=0에서의 초기조건은 ODE 화면에서 확인하십시오. 역변환이 DiracDelta 같은 분포이면 일반 함수 그래프로 바꾸지 않고 기호 결과만 제공합니다. 미평가 변환이나 실수 배열로 변환할 수 없는 결과도 그래프를 생략합니다.

## 5. ODE와 연결
ODE 화면에서 전달한 Y(s)를 역변환하여 시간 응답과 비교하십시오. 초기조건 항 m(s y₀+v₀)+c y₀를 빠뜨리지 않는 것이 중요합니다. 1/(s+1)², 1/(s²+4), exp(−2s)/s를 각각 입력하여 중근, 진동, 시간 지연을 비교하십시오.

## 라플라스 코드별 역할

`LaplaceTopic.compute`는 정변환에서 변환식·수렴 경계·추가 조건을 받습니다. 역변환에서는 s를 복소 변수로 해석하고 DiracDelta 또는 미평가 변환이 포함되는지 검사합니다. 분포를 0 배열로 바꾸지 않습니다. 그래프를 만들 수 없어도 기호 표를 보존합니다. 시간 그래프의 t>0 표본과 초기값의 오른쪽 극한을 구별하십시오.


````python
from engineering_math.topics.laplace import LaplaceTopic, MODES as LAPLACE_MODES
laplace = LaplaceTopic()
forward_result = laplace.compute({'expression': 'sin(2*t)'})
for label, value in forward_result.tables['기호 변환'][1]:
    print(label, ':', value)
inverse_result = laplace.compute({'mode': LAPLACE_MODES[1], 'expression': ode_result.data['laplace_input']})
print('ODE의 Y(s):', ode_result.data['laplace_input'])
print('역변환:', inverse_result.data['time_expression'])
show(laplace.figures(inverse_result)['시간 영역'])
````

## 초기조건과 임펄스 확인

m=2, k=8, c=0, y₀=3, v₀=4에서는 3cos(2t)+2sin(2t)가 예상 해입니다. 초기조건 항을 빠뜨리지 않았는지 검사하십시오. 상수 1의 역변환은 일반 상수 함수가 아니라 임펄스입니다.


````python
initial_value_result = ode.compute({'mass': 2., 'stiffness': 8., 'damping': 0., 'position': 3., 'velocity': 4.})
linked = laplace.compute({'mode': LAPLACE_MODES[1], 'expression': initial_value_result.data['laplace_input']})
positive_t = linked.data['time']
np.testing.assert_allclose(linked.data['values'], 3*np.cos(2*positive_t)+2*np.sin(2*positive_t), atol=1e-12)
impulse = laplace.compute({'mode': LAPLACE_MODES[1], 'expression': '1'})
print('1의 역변환:', impulse.data['time_expression'])
print(impulse.notices[-1])
assert 'values' not in impulse.data
````

기호 계산 탭에서 F(s), Y(s), 초기조건이 반영된 식을 확인하십시오. 직접 입력의 기본값은 수치 계산만이며 기호 계산 옵션을 바꾸면 라플라스 변환도 시도합니다. 미평가 결과나 변환 실패가 있어도 완료된 수치 결과는 유지합니다. 기호 계산이 오래 걸려 취소한 작업은 결과를 제출하지 않습니다. 연결 버튼은 역변환 화면에 전달할 수 있는 Y(s)를 얻었을 때 활성화됩니다.

# 11.3 강제진동 · 22쪽

## 주기 외력의 분해

# 11.6 직교급수와 일반화된 Fourier 급수 · 35쪽

## 일반화된 계수
가중내적 ⟨f,g⟩ᵣ=∫f(x)g(x)r(x)dx를 정의합니다.
직교기저 φₙ의 계수는 cₙ=⟨f,φₙ⟩ᵣ/⟨φₙ,φₙ⟩ᵣ입니다.
부분합 Σcₙφₙ은 같은 기저 공간에서 가중 제곱오차를 최소화합니다.

## 지원 기저
Legendre는 [-1,1], r=1, Pₙ(x), λ=n(n+1)을 사용합니다. L 입력은 이 모드의 구간에 영향을 주지 않습니다.
Bessel은 [0,L], r=x, Jν(jν,n x/L)을 사용합니다. ν는 0~5의 정수이며 jν,n은 양의 영점입니다.
Bessel 전개는 x=L에서 0인 기저를 사용하므로 원함수 끝점이 0이 아니면 끝점 오차가 남습니다.
여기서 N은 최고 다항식 차수가 아니라 포함하는 기저의 개수입니다. Legendre N=3은 P₀,P₁,P₂입니다.

## 검산과 코드
x²=(1/3)P₀+(2/3)P₂를 확인하십시오.
orthogonal_basis는 SciPy의 Legendre 평가·Bessel 영점과 함수를 사용합니다.
project_basis는 사다리꼴 가중치를 이용해 계수·Gram 행렬·누적 부분합을 함께 계산합니다.
큰 N에서 Gram 오차가 커지면 표본 수를 늘리십시오. 이는 직교성 정리의 실패가 아니라 수치적분 해상도 문제일 수 있습니다.

# 12.1 기본개념 · 73쪽

## PDE와 해의 의미
편미분방정식은 여러 독립변수에 대한 편도함수를 포함합니다. 해의 유일성을 논하려면 방정식과 함께 초기조건·경계조건·정의역을 지정해야 합니다.
선형 방정식의 제차 해는 중첩할 수 있으나 비제차 해의 합이 같은 비제차 방정식의 해가 되는 것은 아닙니다.

## 지원 방정식과 검산
- 열: u_t−αu_xx=0 (포물형)
- 파동: u_tt−c²u_xx=0 (쌍곡형)
- Laplace: u_xx+u_yy=0 (타원형)
- Poisson: u_xx+u_yy=f (타원형)

후보 u(x,y,t)를 미분하여 좌변과 우변의 차이인 잔차를 계산합니다. 열·파동 모드에서 y는 미분 연산에 참여하지 않습니다.
잔차가 기호적으로 0이면 식의 정의역에서 PDE를 만족함을 확인할 수 있습니다. 특정 시간의 표본에서 잔차가 0인 것만으로 전체 해임을 증명할 수는 없습니다.
예를 들어 u=t²는 t=0에서 u_t=0이지만 모든 시간의 열방정식 해는 아닙니다.

## 코드의 역할과 관찰
PdeBasicsTopic은 안전한 수식 파서로 후보를 읽고 SymPy의 diff·simplify로 잔차를 구합니다. 수치 그림은 별도로 evaluate로 계산합니다.

분포 미분을 포함하는 sign·불연속 Piecewise 등은 일반 함수 수치 검산 대상이 아닐 수 있습니다.

## 물리 모델
작은 횡진동, 일정 장력 T, 일정 선밀도 ρ, 무감쇠·무외력을 가정하면 c²=T/ρ이고 u_tt=c²u_xx입니다.
이 화면에서는 c를 직접 입력합니다. x는 현을 따라 오른쪽으로 증가하며 u는 기준 직선으로부터의 부호 있는 횡변위입니다.
고정단 조건은 u(0,t)=u(ell,t)=0입니다. 초기 변위 f(x)와 초기 속도 g(x)를 각각 입력하십시오.

## 변수분리
u=X(x)G(t)를 대입하면 고정단 공간 고유함수 sin(nπx/ell), 각주파수 ω_n=cnπ/ell을 얻습니다.
각 모드는 A_n cos(ω_n t)+(V_n/ω_n)sin(ω_n t)로 변하며 A_n과 V_n은 f·g의 사인 계수입니다.
초기조건이 경계와 다르면 끝점의 값이 그대로 재현되지 않습니다.

## 에너지와 검산
단위 선밀도 에너지는 E=1/2∫(u_t²+c²u_x²)dx입니다. 모드 직교성을 이용한 에너지는 시간에 따라 보존됩니다.
f=sin(πx/ell), g=0이면 u=sin(πx/ell)cos(cπt/ell)입니다.
비교 열방정식은 같은 초기 모양을 사용하지만 지수 감쇠하므로 진동과 확산을 구별할 수 있습니다.
